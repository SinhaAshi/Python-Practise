```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df1 = pd.read_csv("Coffee.csv")

df1.head()
```

```python
print("1. Shape of dataset:", df1.shape)
```

```python
print("\n2. Column names:")
print(df1.columns)
```

```python
print("\n3. Data types:")
print(df1.dtypes)
```

```python
print("\n4. First 5 rows:")
print(df1.head())
```

```python
print("\n5. Last 5 rows:")
print(df1.tail())
```

```python
print("\n6. Missing values in each column:")
print(df1.isnull().sum())
```

```python
print("\n7. Number of unique countries:")
print(df1['Country'].nunique())
```

```python
print("\n8. Records per year:")
print(df1['Year'].value_counts().sort_index())
```

```python
print("\n9. Records per country:")
print(df1['Country'].value_counts())
```

```python
print("\n10. Production statistics:")
print("Minimum:", df1['Total Production'].min())
print("Maximum:", df1['Total Production'].max())
print("Average:", df1['Total Production'].mean())
```

```python
print("\n11. Country with highest production record:")
highest = df1.loc[df1['Total Production'].idxmax()]
print(highest[['Country','Year','Total Production']])
```

```python
print("\n12. Rows with missing or zero production:")
missing_prod = df1[df1['Total Production'].isnull() | (df1['Total Production'] == 0)]
print(missing_prod)
```

```python
print("\n13. Average Import Price:")
print(df1['Import Price'].mean())
```

```python
print("\n13. Average Import Price:")
print(df1['Import Price'].mean())
```

```python
print("\n15. Top 5 countries by production:")
print(df1.groupby('Country')['Total Production'].sum().sort_values(ascending=False).head())
```

```python
print("\n16. Country with highest average export price:")
print(df1.groupby('Country')['Export Price'].mean().sort_values(ascending=False).head(1))
```

```python
print("\n17. Rows where domestic consumption > production:")
print(df1[df1['Domestic Consumption'] > df1['Total Production']]
      [['Country','Year','Domestic Consumption','Total Production']])
```

```python
print("\n18. Country with highest production growth:")
growth = df1.groupby('Country')['Total Production'].apply(lambda x: x.max() - x.min())
print(growth.sort_values(ascending=False).head())
```

```python
print("\n19. Correlation between Export Price and Total Production:")
print(df1['Export Price'].corr(df1['Total Production']))
```

```python
print("\n20. Key Insights:")
print("""
1. Production varies significantly between countries.
2. A few countries dominate global production.
3. Export price has only a weak correlation with production.
4. Some countries consume more coffee than they produce.
5. Production trends generally increase over the years.
""")
```

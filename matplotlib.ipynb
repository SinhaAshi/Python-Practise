```python
x = np.arange(1,11)
y = np.sin(x)
z = np.cos(x)

print(x,y,z, sep = '\n\n')
```

```python
plt.plot(x,y)
```

```python
plt.plot(x,y, label = 'SINQ')
plt.plot(y,z, label = 'COSQ')
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (6,4))
plt.plot(x,y,label = 'SINQ', color = 'red')
plt.plot(y,z, label ='COSQ')
plt.title('SINQ and COSQ')
plt.xlabel("X-axis label")
plt.ylabel("Y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y, color='red', linestyle='-', alpha = 0.3,marker = "*", markerfacecolor='yellow', markersize = 10, markeredgecolor='blue', linewidth = 10)
plt.show()
```

```python
plt.figure(figsize = (6,4))

plt.subplot(1,2,1)
plt.title('SINQ and COSQ')
plt.plot(x,y,label = 'SINQ', color = 'red')
plt.plot(y,z, label ='COSQ')
plt.xlabel("Data Points")
plt.ylabel("Associate values")

plt.subplot(1,2,2)
plt.title('SINQ and COSQ')
plt.plot(x,y,label = 'SINQ', color = 'red')
plt.plot(y,z, label ='COSQ')
plt.xlabel("Data Points2")
plt.ylabel("Associate values2")

plt.legend()
plt.show()
```

```python
plt.figure(figsize = (8,4))

plt.subplot(3,3,1)
plt.title('SINQ and COSQ')
plt.plot(x,y,label = 'SINQ', color = 'red')
plt.plot(y,z, label ='COSQ')
plt.xlabel("Data Points")
plt.ylabel("Associate values")

plt.subplot(3,3,5)
plt.title('SINQ and COSQ')
plt.plot(x,y,label = 'SINQ', color = 'red')
plt.plot(y,z, label ='COSQ')
plt.xlabel("Data Points2")
plt.ylabel("Associate values2")

plt.subplot(3,3,9)
plt.title('SINQ and COSQ')
plt.plot(x,y,label = 'SINQ', color = 'red')
plt.plot(y,z, label ='COSQ')
plt.xlabel("Data Points2")
plt.ylabel("Associate values2")
plt.grid()
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 8))

x = df.index
y = df['tip']
z = df['total_bill']

# 1️⃣ Line Plot
plt.subplot(3,3,1)
plt.title('Tips vs Total Bill (Line)')
plt.plot(x, y, label='Tip')
plt.plot(x, z, label='Total Bill')
plt.legend()
plt.grid()

# 2️⃣ Pie Chart (Grouped Data)
plt.subplot(3,3,3)
day_total = df.groupby('day')['total_bill'].sum()
plt.title('Revenue by Day')
plt.pie(day_total, labels=day_total.index, autopct='%1.1f%%')

# 3️⃣ Scatter Plot
plt.subplot(3,3,5)
plt.title('Tip vs Total Bill (Scatter)')
plt.scatter(z, y)
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.grid()

# 4️⃣ Bar Plot (Average Tip by Day)
plt.subplot(3,3,7)
avg_tip = df.groupby('day')['tip'].mean()
plt.title('Avg Tip by Day')
plt.bar(avg_tip.index, avg_tip)
plt.grid()

# 5️⃣ Horizontal Bar (Total Revenue by Day)
plt.subplot(3,3,9)
plt.title('Total Revenue by Day')
plt.barh(day_total.index, day_total)
plt.grid()

plt.tight_layout()
plt.show()
```
